#### Chart - 6

````python
# Chart - 6 visualization code
import pandas as pd
import matplotlib.pyplot as plt

# Convert dates
data['Scheduled Delivery Date'] = pd.to_datetime(
    data['Scheduled Delivery Date'], errors='coerce'
)

data['Delivered to Client Date'] = pd.to_datetime(
    data['Delivered to Client Date'], errors='coerce'
)

# Calculate delivery delay in days
data['Delivery Delay Days'] = (
    data['Delivered to Client Date'] - data['Scheduled Delivery Date']
).dt.days

# Remove missing values
delay_data = data.dropna(subset=['Delivery Delay Days'])

# Create delay categories
delay_data['Delivery Status'] = delay_data['Delivery Delay Days'].apply(
    lambda x: 'Early' if x < 0
    else 'On Time' if x == 0
    else 'Late'
)

# Count delivery status
status_counts = delay_data['Delivery Status'].value_counts()

# Visualization
plt.figure(figsize=(8, 5))

status_counts.plot(kind='bar')

plt.title('Delivery Performance: Early, On-Time and Late')
plt.xlabel('Delivery Status')
plt.ylabel('Number of Shipments')
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()


````

##### 1. Why did you pick the specific chart?

we chose a bar chart for chart because it is the  best way to compare the number of shipment in three delivery categories : EARLY: ON time and LATE

##### 2. What is/are the insight(s) found from the chart?

#### Chart - 13

````python
# Chart - 13 visualization code
# average Delivery Deley by Shipment mode
import pandas as pd
import matplotlib.pyplot as plt

# Convert date columns into datetime format
data['Scheduled Delivery Date'] = pd.to_datetime(
    data['Scheduled Delivery Date'],
    errors='coerce'
)

data['Delivered to Client Date'] = pd.to_datetime(
    data['Delivered to Client Date'],
    errors='coerce'
)

# Calculate delivery delay in days
data['Delivery Delay Days'] = (
    data['Delivered to Client Date']
    - data['Scheduled Delivery Date']
).dt.days

# Calculate average delay by shipment mode
avg_delay = data.groupby('Shipment Mode')['Delivery Delay Days'].mean()
avg_delay = avg_delay.sort_values(ascending=False)

# Create the chart
plt.figure(figsize=(8, 5))

avg_delay.plot(kind='bar')

plt.title('Average Delivery Delay by Shipment Mode')
plt.xlabel('Shipment Mode')
plt.ylabel('Average Delivery Delay (Days)')
plt.xticks(rotation=0)

plt.axhline(0, color='black', linewidth=0.8)

plt.tight_layout()
plt.show()
````

#### Chart - 14 - Correlation Heatmap

````python
# Correlation Heatmap visualization code
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Select important numerical columns
numeric_columns = [
    'Line Item Quantity',
    'Line Item Value',
    'Pack Price',
    'Unit Price',
    'Weight (Kilograms)',
    'Freight Cost (USD)',
    'Line Item Insurance (USD)'
]

# Convert selected columns into numeric format
for column in numeric_columns:
    data[column] = pd.to_numeric(data[column], errors='coerce')

# Calculate correlation
correlation_matrix = data[numeric_columns].corr()

# Create correlation heatmap
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5,
    center=0
)

plt.title('Correlation Heatmap of Numerical Variables')
plt.tight_layout()
plt.show()
````

#### Chart - 15 - Pair Plot

````python
# Pair Plot visualization code
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Select important numerical columns
pair_columns = [
    'Line Item Quantity',
    'Line Item Value',
    'Pack Price',
    'Unit Price',
    'Weight (Kilograms)'
]

# Convert columns into numeric format
for column in pair_columns:
    data[column] = pd.to_numeric(data[column], errors='coerce')

# Remove missing values
pair_data = data[pair_columns].dropna()

# Create pair plot
sns.pairplot(
    pair_data,
    diag_kind='hist',
    corner=True
)

plt.suptitle(
    'Pair Plot of Numerical Supply Chain Variables',
    y=1.02
)

plt.show()
````
